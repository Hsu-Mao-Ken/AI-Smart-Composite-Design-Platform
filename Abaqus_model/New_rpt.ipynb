```python
from abaqus import *
from odbAccess import *
from abaqusConstants import *
from odbMaterial import *
from odbSection import *
import math
import visualization
import odbAccess

modelName = 'test_2_0_1'
fileName = modelName + '.odb'
resultODB = visualization.openOdb(path=fileName, readOnly=True)
Plasticstep = resultODB.steps.values()[0]

matrix_parameters = {'strFxe0x':[], 
                    'strFxe0y':[], 
                    'strFxe0z':[], 
                    'strFye0x':[], 
                    'strFye0y':[], 
                    'strFye0z':[], 
                    'strFze0x':[], 
                    'strFze0y':[], 
                    'strFze0z':[], 
                    'strSyz_g0':[], 
                    'strSzx_g0':[], 
                    'strSxy_g0':[]}


MaterialProperties = []
Time = []
for i in range(len(resultODB.steps[Plasticstep.name].frames)):
    frameFx = resultODB.steps[Plasticstep.name].frames[i]
    Time.append(frameFx.frameValue)
    Fx_eps0_x = frameFx.fieldOutputs['U'].values[0].data[0]
    
    matrix_parameters['strFxe0x'].append(Fx_eps0_x)


reportFilename = modelName + '.rpt'
fOutputReport = open(reportFilename, 'w')

MaterialProperties_data = []
Time_data = []
for i in range(len(Time)):
    Time_string = [Time[i]]
    MaterialProperties_string = [matrix_parameters['strFxe0x'][i]]
    
    Time_data.append(''.join(map(str, Time_string)))
    MaterialProperties_data.append(''.join(map(str, MaterialProperties_string)))
    
    fOutputReport.write(Time_data[i])
    fOutputReport.write('\n')
    fOutputReport.write(MaterialProperties_data[i])
    fOutputReport.write('\n')
    
    
fOutputReport.close()



#getInputs(fields=fieldsMaterialProperties,
        #label='',
        #dialogTitle='Effective material properties obtained from the unit cell analysis',)
```
