```python
#def刪除原本inp檔中用不到的部分
#透過replace簡單的替換材料性質
import os
def delete_content(file1, file2, key_word):
    """
    將替換的字串寫到一個新的檔案中，然後將原檔案刪除，新檔案改為原來檔案的名字
    :param file: 檔案路徑
    :param old_str: 需要替換的字串
    :param new_str: 替換的字串
    :return: None
    """
    data = ''
    ori_mat = '*Elastic\n2.5e+10, 0.2'
    new_mat = '*Elastic\n2.5e+10, 0.2\n*Plastic\n4e+06,  0.\n2e+08, 0.3'
    node = '*NODE\n'
    with open(file1, "r", encoding="windows-1252") as f1, open(file2, "w+", encoding="utf-8") as f2:
        for line in f1.readlines():
            data += line
            if(line.find(key_word) == 0):
                break
                
        f2.write(data.replace(ori_mat, new_mat))
        f2.write(node)
```

```python
import os
import math
def distance(file, key_word_1, key_word_2):
    data1 = ''
    data2 = ''
    with open(file, "r", encoding="windows-1252") as f:
        for line in f.readlines():
            if(line.find(key_word_1) == 0):
                data1 += line
                data1 = data1.replace('\n', '')
                x1 = data1.split(', ')[1]
                y1 = data1.split(', ')[2]
                z1 = data1.split(', ')[3]
                f.seek(0, 0)
                break
        for line in f.readlines():       
            if(line.find(key_word_2) == 0):
                data2 += line
                data2 = data2.replace('\n', '')
                x2 = data2.split(', ')[1]
                y2 = data2.split(', ')[2]
                z2 = data2.split(', ')[3]
                break
        len_x = float(x2) - float(x1)
        len_y = float(y2) - float(y1)
        len_z = float(z2) - float(z1)
        
        # 0.05%
        displacement = math.floor(len_x * 5)/10000
                
    return displacement
```

```python
for i in range(1, 501):
    angle = ((i-1)//100)*15
    file1 = 'Plastic_'+name+'_'+str(angle)+'_'+ str(i) + '.inp'
    displacement = str(distance(file1, '1,', '51,'))
    with open(file1, "rt") as f:
        x = f.read()
    
    with open(file1, "wt") as f:
        x = x.replace("dis_change", displacement)
        f.write(x)
```
